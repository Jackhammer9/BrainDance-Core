Combine And Manage all Data

```python
import os
import numpy as np

RAW_DIR = "raw"
DATASET_DIR = "datasets"
DATASET_FILE = os.path.join(DATASET_DIR, "raw_eye_data.npz")


def load_npz_safe(path):
    data = np.load(path)
    if "X" not in data or "y" not in data:
        raise ValueError(f"{path} does not contain X and y")
    return data["X"], data["y"]


def manage_raw_data():
    os.makedirs(DATASET_DIR, exist_ok=True)

    X_list = []
    y_list = []

    # 1. Load existing dataset if it exists
    if os.path.exists(DATASET_FILE):
        X_old, y_old = load_npz_safe(DATASET_FILE)
        X_list.append(X_old)
        y_list.append(y_old)

    # 2. Load all raw/*.npz files
    raw_files = [
        os.path.join(RAW_DIR, f)
        for f in os.listdir(RAW_DIR)
        if f.endswith(".npz")
    ]

    if not raw_files:
        print("No raw files to process.")
        return

    for f in raw_files:
        X, y = load_npz_safe(f)
        X_list.append(X)
        y_list.append(y)

    # 3. Concatenate
    X_all = np.concatenate(X_list, axis=0)
    y_all = np.concatenate(y_list, axis=0)

    # 4. Save merged dataset
    np.savez(
        DATASET_FILE,
        X=X_all,
        y=y_all
    )

    # 5. Delete all raw files
    for f in raw_files:
        os.remove(f)

    print("Data management complete.")
    print("Final dataset:", DATASET_FILE)
    print("Total samples:", len(y_all))


if __name__ == "__main__":
    manage_raw_data()
```
