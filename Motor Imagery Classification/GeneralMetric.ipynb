```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from itertools import combinations

from scipy.signal import welch, butter, filtfilt, detrend
from scipy.stats import skew, kurtosis, entropy, wasserstein_distance, ks_2samp
from tqdm.auto import tqdm


# ============================================================
# CONFIG
# ============================================================

FS = 250

LEFT_LABEL = 1
RIGHT_LABEL = 2

CLASS_NAMES = {
    LEFT_LABEL: "left",
    RIGHT_LABEL: "right"
}

# Name them however you want.
# These names appear in the matrices and plots.
DATASETS = {
    "Arnav": "Datasets/Main.npz",
    "Fauzaan": "Datasets/Main1.npz",
    "Devin": "Datasets/Main2.npz",
    "Erim" : "Datasets/Main3.npz",
}

BAD_CHANNELS = [0, 1, 2, 4, 6, 7]

FEATURE_DATABASE_PATH = "Datasets/eeg_feature_database.csv"

SAVE_PLOTS = False

LEFT_HEATMAP_PATH = "Datasets/left_similarity_heatmap.png"
RIGHT_HEATMAP_PATH = "Datasets/right_similarity_heatmap.png"

LEFT_BARPLOT_PATH = "Datasets/left_average_similarity.png"
RIGHT_BARPLOT_PATH = "Datasets/right_average_similarity.png"

# If None, remaining channels after deletion become ch0, ch1, ch2...
CHANNEL_NAMES = None

# Speed switches
USE_SLOW_FEATURES = False
USE_CROSS_CHANNEL_BAND_FEATURES = True

# If True, rebuild features from raw npz files.
# If False, reload existing eeg_feature_database.csv.
REBUILD_FEATURE_DATABASE = True

# ============================================================
# IMPORTANT SCORE KNOB
# ============================================================
# Similarity is computed from the top K most similar features per person-pair/class.
# Example:
# TOP_K_FEATURES_FOR_SCORE = 10  -> only use top 10 most similar features
# TOP_K_FEATURES_FOR_SCORE = 50  -> use top 50
# TOP_K_FEATURES_FOR_SCORE = None -> use all features
TOP_K_FEATURES_FOR_SCORE = 50

EPS = 1e-12

BANDS = {
    "delta": (1, 4),
    "theta": (4, 8),
    "alpha": (8, 13),
    "low_beta": (13, 20),
    "high_beta": (20, 30),
    "beta": (13, 30),
    "gamma_low": (30, 45),
    "mu": (8, 13),
    "smr": (12, 15),
}


# ============================================================
# BASIC UTILS
# ============================================================

def safe(x):
    if np.isnan(x) or np.isinf(x):
        return 0.0
    return float(x)


def bandpass(sig, fs, low=1, high=45, order=4):
    nyq = fs / 2
    b, a = butter(order, [low / nyq, high / nyq], btype="band")
    return filtfilt(b, a, sig)


def bandpower(sig, fs, low, high):
    freqs, psd = welch(
        sig,
        fs=fs,
        nperseg=min(256, len(sig)),
        noverlap=min(128, len(sig) // 2)
    )

    mask = (freqs >= low) & (freqs <= high)

    if not np.any(mask):
        return 0.0

    return safe(np.trapezoid(psd[mask], freqs[mask]))


def spectral_entropy(sig, fs):
    freqs, psd = welch(
        sig,
        fs=fs,
        nperseg=min(256, len(sig)),
        noverlap=min(128, len(sig) // 2)
    )

    psd = psd + EPS
    psd_norm = psd / np.sum(psd)

    return safe(entropy(psd_norm) / np.log(len(psd_norm)))


def hjorth_params(sig):
    first_deriv = np.diff(sig)
    second_deriv = np.diff(first_deriv)

    var0 = np.var(sig) + EPS
    var1 = np.var(first_deriv) + EPS
    var2 = np.var(second_deriv) + EPS

    activity = var0
    mobility = np.sqrt(var1 / var0)
    complexity = np.sqrt(var2 / var1) / (mobility + EPS)

    return safe(activity), safe(mobility), safe(complexity)


def zero_crossing_rate(sig):
    return safe(np.mean(np.diff(np.signbit(sig))))


def line_length(sig):
    return safe(np.sum(np.abs(np.diff(sig))))


def rms(sig):
    return safe(np.sqrt(np.mean(sig ** 2)))


def mean_abs(sig):
    return safe(np.mean(np.abs(sig)))


def energy(sig):
    return safe(np.sum(sig ** 2))


def shannon_entropy_hist(sig, bins=30):
    hist, _ = np.histogram(sig, bins=bins, density=True)
    hist = hist + EPS
    hist = hist / np.sum(hist)
    return safe(entropy(hist))


def approximate_entropy(sig, m=2, r=None):
    sig = np.asarray(sig)

    if r is None:
        r = 0.2 * np.std(sig)

    N = len(sig)

    if N <= m + 1:
        return 0.0

    def _phi(mm):
        x = np.array([sig[i:i + mm] for i in range(N - mm + 1)])
        C = []

        for xi in x:
            dist = np.max(np.abs(x - xi), axis=1)
            C.append(np.mean(dist <= r))

        C = np.array(C) + EPS
        return np.mean(np.log(C))

    return safe(_phi(m) - _phi(m + 1))


def higuchi_fd(sig, kmax=8):
    sig = np.asarray(sig)
    N = len(sig)

    if N < kmax * 2:
        return 0.0

    L = []
    k_values = np.arange(1, kmax + 1)

    for k in k_values:
        Lk = []

        for m in range(k):
            idx = np.arange(m, N, k)

            if len(idx) < 2:
                continue

            dist = np.sum(np.abs(np.diff(sig[idx])))
            norm = (N - 1) / (len(idx) * k)
            Lk.append(dist * norm)

        L.append(np.mean(Lk) + EPS)

    L = np.array(L)
    coeffs = np.polyfit(np.log(1.0 / k_values), np.log(L), 1)

    return safe(coeffs[0])


def spectral_edge_frequency(sig, fs, edge=0.95):
    freqs, psd = welch(
        sig,
        fs=fs,
        nperseg=min(256, len(sig)),
        noverlap=min(128, len(sig) // 2)
    )

    cumulative = np.cumsum(psd)
    total = cumulative[-1] + EPS

    idx = np.searchsorted(cumulative, edge * total)

    if idx >= len(freqs):
        idx = len(freqs) - 1

    return safe(freqs[idx])


def dominant_frequency(sig, fs):
    freqs, psd = welch(
        sig,
        fs=fs,
        nperseg=min(256, len(sig)),
        noverlap=min(128, len(sig) // 2)
    )

    idx = np.argmax(psd)
    return safe(freqs[idx])


# ============================================================
# FEATURE EXTRACTION
# ============================================================

def extract_features_for_trial(trial, fs):
    """
    trial shape: (channels, samples)
    returns dict of features
    """

    features = {}

    n_channels = trial.shape[0]

    if CHANNEL_NAMES is None:
        ch_names = [f"ch{i}" for i in range(n_channels)]
    else:
        ch_names = CHANNEL_NAMES

    if len(ch_names) != n_channels:
        raise ValueError(
            f"CHANNEL_NAMES length {len(ch_names)} does not match number of channels {n_channels}"
        )

    filtered_trial = []

    for ch_idx, ch_name in enumerate(ch_names):
        sig_raw = trial[ch_idx].astype(float)

        sig = detrend(sig_raw)
        sig = bandpass(sig, fs, 1, 45)

        filtered_trial.append(sig)

        features[f"{ch_name}_mean"] = safe(np.mean(sig))
        features[f"{ch_name}_std"] = safe(np.std(sig))
        features[f"{ch_name}_var"] = safe(np.var(sig))
        features[f"{ch_name}_logvar"] = safe(np.log(np.var(sig) + EPS))
        features[f"{ch_name}_median"] = safe(np.median(sig))
        features[f"{ch_name}_min"] = safe(np.min(sig))
        features[f"{ch_name}_max"] = safe(np.max(sig))
        features[f"{ch_name}_ptp"] = safe(np.ptp(sig))
        features[f"{ch_name}_rms"] = rms(sig)
        features[f"{ch_name}_mean_abs"] = mean_abs(sig)
        features[f"{ch_name}_energy"] = energy(sig)
        features[f"{ch_name}_line_length"] = line_length(sig)
        features[f"{ch_name}_zcr"] = zero_crossing_rate(sig)
        features[f"{ch_name}_skew"] = safe(skew(sig))
        features[f"{ch_name}_kurtosis"] = safe(kurtosis(sig))
        features[f"{ch_name}_iqr"] = safe(np.percentile(sig, 75) - np.percentile(sig, 25))

        activity, mobility, complexity = hjorth_params(sig)

        features[f"{ch_name}_hjorth_activity"] = activity
        features[f"{ch_name}_hjorth_mobility"] = mobility
        features[f"{ch_name}_hjorth_complexity"] = complexity

        features[f"{ch_name}_spectral_entropy"] = spectral_entropy(sig, fs)
        features[f"{ch_name}_hist_entropy"] = shannon_entropy_hist(sig)

        if USE_SLOW_FEATURES:
            features[f"{ch_name}_approx_entropy"] = approximate_entropy(sig)
            features[f"{ch_name}_higuchi_fd"] = higuchi_fd(sig)

        total_power = bandpower(sig, fs, 1, 45) + EPS

        band_powers = {}

        for band_name, (low, high) in BANDS.items():
            bp = bandpower(sig, fs, low, high)
            band_powers[band_name] = bp + EPS

            features[f"{ch_name}_{band_name}_power"] = bp
            features[f"{ch_name}_{band_name}_rel_power"] = safe(bp / total_power)
            features[f"{ch_name}_{band_name}_log_power"] = safe(np.log(bp + EPS))

        features[f"{ch_name}_dominant_freq"] = dominant_frequency(sig, fs)
        features[f"{ch_name}_spectral_edge_95"] = spectral_edge_frequency(sig, fs, 0.95)
        features[f"{ch_name}_spectral_edge_90"] = spectral_edge_frequency(sig, fs, 0.90)

        alpha = band_powers["alpha"]
        beta = band_powers["beta"]
        theta = band_powers["theta"]
        delta = band_powers["delta"]
        gamma = band_powers["gamma_low"]
        mu = band_powers["mu"]
        smr = band_powers["smr"]

        features[f"{ch_name}_alpha_beta_ratio"] = safe(alpha / beta)
        features[f"{ch_name}_theta_beta_ratio"] = safe(theta / beta)
        features[f"{ch_name}_alpha_theta_ratio"] = safe(alpha / theta)
        features[f"{ch_name}_delta_theta_ratio"] = safe(delta / theta)
        features[f"{ch_name}_beta_alpha_ratio"] = safe(beta / alpha)
        features[f"{ch_name}_gamma_beta_ratio"] = safe(gamma / beta)
        features[f"{ch_name}_mu_beta_ratio"] = safe(mu / beta)
        features[f"{ch_name}_smr_beta_ratio"] = safe(smr / beta)

    filtered_trial = np.array(filtered_trial)

    if n_channels >= 2:
        for i in range(n_channels):
            for j in range(i + 1, n_channels):
                ch_i = ch_names[i]
                ch_j = ch_names[j]

                sig_i = filtered_trial[i]
                sig_j = filtered_trial[j]

                features[f"{ch_i}_{ch_j}_corr"] = safe(np.corrcoef(sig_i, sig_j)[0, 1])
                features[f"{ch_i}_{ch_j}_cov"] = safe(np.cov(sig_i, sig_j)[0, 1])

                if USE_CROSS_CHANNEL_BAND_FEATURES:
                    for band_name, (low, high) in BANDS.items():
                        bi = bandpass(sig_i, fs, low, high)
                        bj = bandpass(sig_j, fs, low, high)

                        features[f"{ch_i}_{ch_j}_{band_name}_corr"] = safe(
                            np.corrcoef(bi, bj)[0, 1]
                        )

                        features[f"{ch_i}_{ch_j}_{band_name}_cov"] = safe(
                            np.cov(bi, bj)[0, 1]
                        )

                        power_i = bandpower(sig_i, fs, low, high) + EPS
                        power_j = bandpower(sig_j, fs, low, high) + EPS

                        features[f"{ch_i}_{ch_j}_{band_name}_power_diff"] = safe(
                            power_i - power_j
                        )

                        features[f"{ch_i}_{ch_j}_{band_name}_power_abs_diff"] = safe(
                            abs(power_i - power_j)
                        )

                        features[f"{ch_i}_{ch_j}_{band_name}_power_ratio"] = safe(
                            power_i / power_j
                        )

                        features[f"{ch_i}_{ch_j}_{band_name}_log_power_ratio"] = safe(
                            np.log(power_i / power_j)
                        )

    return features


def extract_dataset_features(X, y, person_name, dataset_path):
    rows = []

    print(f"\n========== EXTRACTING FEATURES: {person_name} ==========")
    print(f"Dataset path: {dataset_path}")
    print(f"X shape: {X.shape}")
    print(f"Labels: {np.unique(y, return_counts=True)}")
    print(f"USE_SLOW_FEATURES: {USE_SLOW_FEATURES}")
    print(f"USE_CROSS_CHANNEL_BAND_FEATURES: {USE_CROSS_CHANNEL_BAND_FEATURES}")

    for trial_idx in tqdm(range(X.shape[0]), desc=f"{person_name} trials"):
        label = y[trial_idx]

        if label not in CLASS_NAMES:
            continue

        feats = extract_features_for_trial(X[trial_idx], FS)

        feats["person"] = person_name
        feats["dataset_path"] = dataset_path
        feats["trial_idx"] = trial_idx
        feats["label"] = label
        feats["class_name"] = CLASS_NAMES.get(label, str(label))

        rows.append(feats)

    df = pd.DataFrame(rows)

    print(f"Done: {person_name}")
    print(f"Feature table shape: {df.shape}")

    return df


# ============================================================
# FEATURE DATABASE
# ============================================================

def load_raw_dataset(path):
    data = np.load(path)

    X = data["X_raw"]
    y = data["y"]

    print("\nRaw loaded:", path)
    print("Before bad-channel deletion:", X.shape)

    X = np.delete(X, BAD_CHANNELS, axis=1)

    print("After bad-channel deletion:", X.shape)

    return X, y


def build_feature_database(datasets):
    all_dfs = []

    print("\n========== BUILDING FEATURE DATABASE ==========")

    os.makedirs(os.path.dirname(FEATURE_DATABASE_PATH), exist_ok=True)

    expected_channels = None

    for person_name, dataset_path in datasets.items():
        print(f"\nLoading {person_name}: {dataset_path}")

        X, y = load_raw_dataset(dataset_path)

        if expected_channels is None:
            expected_channels = X.shape[1]
        else:
            if X.shape[1] != expected_channels:
                raise ValueError(
                    f"Channel mismatch. Previous datasets had {expected_channels} channels, "
                    f"but {person_name} has {X.shape[1]} channels after deletion."
                )

        print(f"{person_name} labels:", np.unique(y, return_counts=True))

        df = extract_dataset_features(
            X=X,
            y=y,
            person_name=person_name,
            dataset_path=dataset_path
        )

        all_dfs.append(df)

    if len(all_dfs) == 0:
        raise ValueError("No datasets were loaded.")

    feature_db = pd.concat(all_dfs, ignore_index=True)

    metadata_cols = [
        "person",
        "dataset_path",
        "trial_idx",
        "label",
        "class_name"
    ]

    feature_cols = [
        c for c in feature_db.columns
        if c not in metadata_cols
    ]

    feature_db = feature_db[metadata_cols + feature_cols]

    feature_db.to_csv(FEATURE_DATABASE_PATH, index=False)

    print("\n========== FEATURE DATABASE SAVED ==========")
    print("Saved:", FEATURE_DATABASE_PATH)
    print("Shape:", feature_db.shape)
    print("People:", feature_db["person"].unique())
    print("Labels:")
    print(feature_db.groupby(["person", "class_name"]).size())

    return feature_db


def load_feature_database(path=FEATURE_DATABASE_PATH):
    print(f"\nLoading feature database: {path}")
    feature_db = pd.read_csv(path)

    print("Shape:", feature_db.shape)
    print("People:", feature_db["person"].unique())
    print("Labels:")
    print(feature_db.groupby(["person", "class_name"]).size())

    return feature_db


def get_feature_columns(feature_db):
    metadata_cols = [
        "person",
        "dataset_path",
        "trial_idx",
        "label",
        "class_name"
    ]

    return [c for c in feature_db.columns if c not in metadata_cols]


# ============================================================
# SIMILARITY METRICS
# ============================================================

def compare_feature_distributions(a, b):
    """
    Higher similarity_score = more similar.
    """

    a = np.asarray(a, dtype=float)
    b = np.asarray(b, dtype=float)

    a = a[np.isfinite(a)]
    b = b[np.isfinite(b)]

    if len(a) < 2 or len(b) < 2:
        return None

    mean_a = np.mean(a)
    mean_b = np.mean(b)

    std_a = np.std(a) + EPS
    std_b = np.std(b) + EPS

    pooled_std = np.sqrt((std_a ** 2 + std_b ** 2) / 2) + EPS

    abs_mean_diff = abs(mean_a - mean_b)
    normalized_mean_diff = abs_mean_diff / pooled_std

    median_diff = abs(np.median(a) - np.median(b)) / pooled_std

    wd = wasserstein_distance(a, b) / pooled_std

    ks_stat, ks_p = ks_2samp(a, b)

    similarity_score = 1.0 / (1.0 + normalized_mean_diff + wd + ks_stat)

    return {
        "person_a_mean": mean_a,
        "person_b_mean": mean_b,
        "person_a_std": std_a,
        "person_b_std": std_b,
        "abs_mean_diff": abs_mean_diff,
        "normalized_mean_diff": normalized_mean_diff,
        "normalized_median_diff": median_diff,
        "wasserstein_norm": wd,
        "ks_stat": ks_stat,
        "ks_p_value": ks_p,
        "similarity_score": similarity_score
    }


def compare_people_for_class(feature_db, person_a, person_b, class_label):
    feature_cols = get_feature_columns(feature_db)

    rows = []

    d1 = feature_db[
        (feature_db["person"] == person_a) &
        (feature_db["label"] == class_label)
    ]

    d2 = feature_db[
        (feature_db["person"] == person_b) &
        (feature_db["label"] == class_label)
    ]

    if len(d1) == 0 or len(d2) == 0:
        print(
            f"Skipping {person_a} vs {person_b}, class {class_label}. "
            f"Trials: {person_a}={len(d1)}, {person_b}={len(d2)}"
        )
        return pd.DataFrame()

    for feat in feature_cols:
        result = compare_feature_distributions(d1[feat], d2[feat])

        if result is None:
            continue

        result["feature"] = feat
        result["person_a"] = person_a
        result["person_b"] = person_b
        result["class_label"] = class_label
        result["class_name"] = CLASS_NAMES.get(class_label, str(class_label))

        rows.append(result)

    out = pd.DataFrame(rows)

    if len(out) == 0:
        return out

    out = out.sort_values("similarity_score", ascending=False).reset_index(drop=True)

    return out


def score_from_similarity_table(similarity_table):
    """
    Computes final person-to-person similarity score.

    Uses TOP_K_FEATURES_FOR_SCORE.
    If TOP_K_FEATURES_FOR_SCORE is None, uses all features.
    """

    if len(similarity_table) == 0:
        return np.nan

    scores = similarity_table["similarity_score"].values

    if TOP_K_FEATURES_FOR_SCORE is None:
        return safe(np.mean(scores))

    k = min(TOP_K_FEATURES_FOR_SCORE, len(scores))

    return safe(np.mean(scores[:k]))


def compute_pairwise_person_similarity(feature_db):
    people = sorted(feature_db["person"].unique())

    pair_rows = []
    detailed_tables = {}

    print("\n========== COMPUTING PAIRWISE PERSON SIMILARITY ==========")
    print("People:", people)
    print("Top K features used for final score:", TOP_K_FEATURES_FOR_SCORE)

    for person_a, person_b in tqdm(list(combinations(people, 2)), desc="Person pairs"):
        print(f"\nComparing {person_a} vs {person_b}")

        left_table = compare_people_for_class(
            feature_db=feature_db,
            person_a=person_a,
            person_b=person_b,
            class_label=LEFT_LABEL
        )

        right_table = compare_people_for_class(
            feature_db=feature_db,
            person_a=person_a,
            person_b=person_b,
            class_label=RIGHT_LABEL
        )

        left_score = score_from_similarity_table(left_table)
        right_score = score_from_similarity_table(right_table)

        pair_rows.append({
            "person_a": person_a,
            "person_b": person_b,
            "left_score": left_score,
            "right_score": right_score,
            "num_left_features": len(left_table),
            "num_right_features": len(right_table),
        })

        detailed_tables[(person_a, person_b, "left")] = left_table
        detailed_tables[(person_a, person_b, "right")] = right_table

        print(f"{person_a} vs {person_b}")
        print(f"LEFT score:  {left_score:.4f}")
        print(f"RIGHT score: {right_score:.4f}")

    pairwise_scores = pd.DataFrame(pair_rows)

    print("\n========== PAIRWISE SCORES ==========")
    print(pairwise_scores)

    return pairwise_scores, detailed_tables


def build_similarity_matrix(pairwise_scores, people, score_col):
    matrix = pd.DataFrame(
        np.eye(len(people)),
        index=people,
        columns=people,
        dtype=float
    )

    for _, row in pairwise_scores.iterrows():
        a = row["person_a"]
        b = row["person_b"]
        score = row[score_col]

        matrix.loc[a, b] = score
        matrix.loc[b, a] = score

    return matrix


def compute_person_average_similarity(similarity_matrix):
    rows = []

    people = list(similarity_matrix.index)

    for person in people:
        others = [p for p in people if p != person]
        scores = similarity_matrix.loc[person, others].values.astype(float)

        rows.append({
            "person": person,
            "average_similarity_to_others": np.nanmean(scores),
            "min_similarity_to_others": np.nanmin(scores),
            "max_similarity_to_others": np.nanmax(scores),
        })

    out = pd.DataFrame(rows)

    out = out.sort_values(
        "average_similarity_to_others",
        ascending=False
    ).reset_index(drop=True)

    return out


# ============================================================
# PLOTTING
# ============================================================

def plot_similarity_heatmap(similarity_matrix, title, save_path=None):
    fig, ax = plt.subplots(figsize=(8, 6))

    im = ax.imshow(similarity_matrix.values, vmin=0, vmax=1)

    ax.set_xticks(np.arange(len(similarity_matrix.columns)))
    ax.set_yticks(np.arange(len(similarity_matrix.index)))

    ax.set_xticklabels(similarity_matrix.columns, rotation=45, ha="right")
    ax.set_yticklabels(similarity_matrix.index)

    ax.set_title(title)

    for i in range(len(similarity_matrix.index)):
        for j in range(len(similarity_matrix.columns)):
            value = similarity_matrix.values[i, j]

            if np.isnan(value):
                text = "nan"
            else:
                text = f"{value:.2f}"

            ax.text(j, i, text, ha="center", va="center")

    fig.colorbar(im, ax=ax, label="Similarity score")

    plt.tight_layout()

    if SAVE_PLOTS and save_path is not None:
        os.makedirs(os.path.dirname(save_path), exist_ok=True)
        plt.savefig(save_path, dpi=200)
        print("Saved heatmap:", save_path)

    plt.show()


def plot_person_average_similarity(person_average_scores, title, save_path=None):
    fig, ax = plt.subplots(figsize=(8, 5))

    ax.bar(
        person_average_scores["person"],
        person_average_scores["average_similarity_to_others"]
    )

    ax.set_ylim(0, 1)
    ax.set_ylabel("Average similarity to others")
    ax.set_title(title)

    for i, value in enumerate(person_average_scores["average_similarity_to_others"]):
        ax.text(i, value, f"{value:.2f}", ha="center", va="bottom")

    plt.xticks(rotation=45, ha="right")
    plt.tight_layout()

    if SAVE_PLOTS and save_path is not None:
        os.makedirs(os.path.dirname(save_path), exist_ok=True)
        plt.savefig(save_path, dpi=200)
        print("Saved bar plot:", save_path)

    plt.show()


# ============================================================
# SHARED FEATURE RANKING
# ============================================================

def rank_features_shared_across_all_people(feature_db, detailed_tables, class_name):
    """
    class_name:
    - "left"
    - "right"
    """

    if class_name not in ["left", "right"]:
        raise ValueError("class_name must be 'left' or 'right'")

    all_tables = []

    for key, table in detailed_tables.items():
        person_a, person_b, cls = key

        if cls != class_name:
            continue

        if len(table) == 0:
            continue

        tmp = table.copy()
        tmp["pair"] = f"{person_a} vs {person_b}"
        all_tables.append(tmp)

    if len(all_tables) == 0:
        return pd.DataFrame()

    all_pairwise = pd.concat(all_tables, ignore_index=True)

    shared = (
        all_pairwise
        .groupby("feature")
        .agg(
            mean_similarity=("similarity_score", "mean"),
            min_similarity=("similarity_score", "min"),
            max_similarity=("similarity_score", "max"),
            std_similarity=("similarity_score", "std"),
            mean_normalized_diff=("normalized_mean_diff", "mean"),
            mean_wasserstein=("wasserstein_norm", "mean"),
            mean_ks=("ks_stat", "mean"),
            num_pairs=("similarity_score", "count")
        )
        .reset_index()
    )

    shared["std_similarity"] = shared["std_similarity"].fillna(0.0)

    shared = shared.sort_values(
        ["mean_similarity", "min_similarity", "std_similarity"],
        ascending=[False, False, True]
    ).reset_index(drop=True)

    return shared


# ============================================================
# MAIN
# ============================================================

if REBUILD_FEATURE_DATABASE or not os.path.exists(FEATURE_DATABASE_PATH):
    feature_db = build_feature_database(DATASETS)
else:
    feature_db = load_feature_database(FEATURE_DATABASE_PATH)


print("\n========== AVAILABLE PEOPLE ==========")
print(feature_db["person"].unique())

print("\n========== DATABASE LABEL COUNTS ==========")
print(feature_db.groupby(["person", "class_name"]).size())


pairwise_scores, detailed_tables = compute_pairwise_person_similarity(feature_db)

people = sorted(feature_db["person"].unique())


# ============================================================
# CLASS-SPECIFIC MATRICES
# ============================================================

left_matrix = build_similarity_matrix(
    pairwise_scores=pairwise_scores,
    people=people,
    score_col="left_score"
)

right_matrix = build_similarity_matrix(
    pairwise_scores=pairwise_scores,
    people=people,
    score_col="right_score"
)

left_average_scores = compute_person_average_similarity(left_matrix)
right_average_scores = compute_person_average_similarity(right_matrix)


print("\n========== LEFT SIMILARITY MATRIX ==========")
print(left_matrix)

print("\n========== RIGHT SIMILARITY MATRIX ==========")
print(right_matrix)

print("\n========== LEFT PER-PERSON AVERAGE SIMILARITY ==========")
print(left_average_scores)

print("\n========== RIGHT PER-PERSON AVERAGE SIMILARITY ==========")
print(right_average_scores)


plot_similarity_heatmap(
    similarity_matrix=left_matrix,
    title=f"LEFT Class EEG Similarity Matrix | Top {TOP_K_FEATURES_FOR_SCORE} Features",
    save_path=LEFT_HEATMAP_PATH
)

plot_similarity_heatmap(
    similarity_matrix=right_matrix,
    title=f"RIGHT Class EEG Similarity Matrix | Top {TOP_K_FEATURES_FOR_SCORE} Features",
    save_path=RIGHT_HEATMAP_PATH
)

plot_person_average_similarity(
    person_average_scores=left_average_scores,
    title=f"LEFT Class Average Similarity | Top {TOP_K_FEATURES_FOR_SCORE} Features",
    save_path=LEFT_BARPLOT_PATH
)

plot_person_average_similarity(
    person_average_scores=right_average_scores,
    title=f"RIGHT Class Average Similarity | Top {TOP_K_FEATURES_FOR_SCORE} Features",
    save_path=RIGHT_BARPLOT_PATH
)


# ============================================================
# SHARED FEATURE TABLES
# ============================================================

shared_left_features = rank_features_shared_across_all_people(
    feature_db=feature_db,
    detailed_tables=detailed_tables,
    class_name="left"
)

shared_right_features = rank_features_shared_across_all_people(
    feature_db=feature_db,
    detailed_tables=detailed_tables,
    class_name="right"
)


print("\n========== TOP SHARED LEFT FEATURES ==========")
print(shared_left_features.head(50))

print("\n========== TOP SHARED RIGHT FEATURES ==========")
print(shared_right_features.head(50))


print("\n========== FINAL RESULT ==========")

print("\nPairwise person similarity:")
print(pairwise_scores)

print("\nLEFT similarity matrix:")
print(left_matrix)

print("\nRIGHT similarity matrix:")
print(right_matrix)

print("\nLEFT average similarity:")
print(left_average_scores)

print("\nRIGHT average similarity:")
print(right_average_scores)

print("\nOnly feature database saved as:")
print(FEATURE_DATABASE_PATH)

if SAVE_PLOTS:
    print("\nPlots saved as:")
    print(LEFT_HEATMAP_PATH)
    print(RIGHT_HEATMAP_PATH)
    print(LEFT_BARPLOT_PATH)
    print(RIGHT_BARPLOT_PATH)
```
